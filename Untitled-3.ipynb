```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
# Чтение данных из CSV файла
df = pd.read_csv('03-09-popitkaprofile2.csv')
y = np.array(df.iloc[:,1])
x = np.array(df.iloc[:,0])
# Построение графика
plt.figure(figsize=(18, 10))
plt.plot(x, y, 's-', color='purple', 
         markersize=8, linewidth=2, markerfacecolor='yellow')

#plt.title('График из данных файла', fontsize=16)
plt.xlabel('Время, с', fontsize=12)
plt.ylabel('Скорость, м/c', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from scipy.interpolate import interp1d
# Чтение данных из CSV файла
df = pd.read_csv('03-09-popitkaprofile2.csv')
y = np.array(df.iloc[:,1])
x = np.array(df.iloc[:,0])
spl = interp1d(x, y, kind='cubic')
x_new = np.linspace(x[0], x[-1], 10000)
y_new = spl(x_new)
plt.figure(figsize=(18, 10))

plt.xlabel('Время, с', fontsize=12)
plt.ylabel('Скорость, м/c', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.plot(x_new, y_new)
plt.scatter(x,y, color = 'k')
plt.xlabel('Время, с', fontsize=12)
plt.ylabel('Скорость, м/c', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from scipy.interpolate import interp1d
from scipy.fft import fft, fftfreq
# Чтение данных из CSV файла
df = pd.read_csv('03-09-popitkaprofile2.csv')
y = np.array(df.iloc[:,1])
x = np.array(df.iloc[:,0])
dt = x[1] - x[0]
fft_signal = np.fft.fft(y)
fft_signal = fft_signal[:len(fft_signal)//2-1]
freq = np.fft.fftfreq(len(y), d = dt)
freq = freq[freq>0]
plt.figure(figsize=(18, 10))
plt.plot(freq, np.abs(fft_signal))
plt.xlabel('частота, Гц')
plt.ylabel('Амплитуда')
plt.title('Спектр')
```
