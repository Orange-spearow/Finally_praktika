```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from scipy.interpolate import interp1d
# Чтение данных из CSV файла
df = pd.read_csv('C304-09-popitkalecroy300000-profile.csv')
y = np.array(df.iloc[:,1])
x = np.array(df.iloc[:,0])
spl = interp1d(x, y, kind='cubic')
x_new = np.linspace(x[0], x[-1], 10000)
y_new = spl(x_new)
x_errors = 5e-6
y_errors = 0.0001848
plt.figure(figsize=(18, 10))

plt.xlabel('Время, с', fontsize=12)
plt.ylabel('Скорость, м/c', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.plot(x_new, y_new)
plt.scatter(x,y, color = 'k')
plt.errorbar(x, y, yerr=y_errors, fmt='none', color='green', 
            capsize=4, capthick=1, alpha=0.7, label='Погрешности V')

# Для погрешностей X 
plt.errorbar(x, y, xerr=x_errors, fmt='none', color='red',
            capsize=4, capthick=1, alpha=0.7, label='Погрешности t ')
plt.xlabel('Время, с', fontsize=12)
plt.ylabel('Скорость, м/c', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import pandas as pd

def linear_function(x, a, b):
    """Линейная функция для аппроксимации"""
    return a * x + b

def curve_fit_with_xy_errors(f, x, y, x_err, y_err, p0=None, max_iter=10, tol=1e-6):
    """
    Аппроксимация с учетом погрешностей по обеим осям (итеративный метод)
    """
    if p0 is None:
        p0 = [1.0, 1.0]
    
    params = p0.copy()
    
    for i in range(max_iter):
        current_slope = params[0]
        total_sigma = np.sqrt(y_err**2 + (current_slope * x_err)**2)
        new_params, pcov = curve_fit(f, x, y, sigma=total_sigma, 
                                    absolute_sigma=True, p0=params)
        
        if np.allclose(params, new_params, rtol=tol):
            break
        params = new_params
    
    return new_params, pcov

# Загрузка данных
df = pd.read_csv(r'C:\Users\LTP\Documents\GitHub\Finally_praktika\СеверСтальВМаге\C304-09-popitkalecroy300000-profile2.csv')
y = np.array(df.iloc[:, 1])
x = np.array(df.iloc[:, 0])

# Погрешности измерений
x_errors = 5e-6
y_errors = 0.0001848

# Аппроксимация с погрешностями X и Y
popt_xy, pcov_xy = curve_fit_with_xy_errors(linear_function, x, y, 
                                           x_errors, y_errors)

# Извлекаем параметры и их погрешности
a_xy, b_xy = popt_xy
sigma_a_xy = np.sqrt(pcov_xy[0, 0])
sigma_b_xy = np.sqrt(pcov_xy[1, 1])

#print("Результаты аппроксимации с учетом погрешностей X и Y:")
#print(f"y = ({a_xy:.3e} ± {sigma_a_xy:.3e})x + ({b_xy:.3e} ± {sigma_b_xy:.3e})")

# Создаем данные для построения прямой
x_fit = np.linspace(min(x), max(x), 100)
y_fit_xy = linear_function(x_fit, a_xy, b_xy)

# Область погрешностей
y_upper_xy = linear_function(x_fit, a_xy + sigma_a_xy, b_xy + sigma_b_xy)
y_lower_xy = linear_function(x_fit, a_xy - sigma_a_xy, b_xy - sigma_b_xy)

# Построение графика
plt.figure(figsize=(12, 8))

# Экспериментальные точки - сначала без погрешностей X для лучшей видимости
plt.plot(x, y, 'o', color='green', markersize=6, label='Экспериментальные точки')

# Отдельно рисуем погрешности Y
plt.errorbar(x, y, yerr=y_errors, fmt='none', color='green', 
            capsize=4, capthick=1, alpha=0.7, label='Погрешности V')

# Для погрешностей X 
plt.errorbar(x, y, xerr=x_errors, fmt='none', color='red',
            capsize=4, capthick=1, alpha=0.7, label='Погрешности t ')

# Линия аппроксимации
plt.plot(x_fit, y_fit_xy, 'blue', linewidth=3, 
        label=f'y = {a_xy:.2e}x + {b_xy:.2e}')

# Область погрешностей
plt.fill_between(x_fit, y_lower_xy, y_upper_xy, color='blue', alpha=0.2,
                label='Область погрешности параметров')

plt.xlabel('Время t, с', fontsize=12)
plt.ylabel('Скорость V, м/c', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)

# Добавляем информацию о погрешностях в аннотацию
textstr = f'Погрешности:\nσ_t = {x_errors:.1e}\nσ_v = {y_errors:.1e}\n'
props = dict(boxstyle='round', facecolor='wheat', alpha=0.8)
plt.text(0.05, 0.95, textstr, transform=plt.gca().transAxes, fontsize=10,
         verticalalignment='top', bbox=props)

plt.tight_layout()
plt.show()

# Дополнительная информация о данных
print(f"\nДополнительная информация:")
print(f"Количество точек: {len(x)}")
print(f"Диапазон X: [{min(x):.3e}, {max(x):.3e}]")
print(f"Диапазон Y: [{min(y):.3e}, {max(y):.3e}]")
print(f"Погрешность X: {x_errors:.1e}")
print(f"Погрешность Y: {y_errors:.1e}")
```
